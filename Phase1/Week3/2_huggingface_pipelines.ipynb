```python
# Create a sentiment analysis pipeline

sentiment_model = pipeline("sentiment-analysis")

# Test it with a sample sentence
result = sentiment_model("I really love how Transformers work!")

print(result)
```

```python
# Create a GPT-2 text generation pipeline
generator = pipeline("text-generation", model="gpt2")

result = generator(
    "Once upon a time in a small village,",
    max_length=50,      # how long the output should be
    num_return_sequences=1  # number of generated sentences
)

print(result[0]["generated_text"])
```

```python
# Create a question-answering pipeline
question_answerer = pipeline("question-answering")

# context to check (already built in model will cheeck on thsi though )
context = """
Transformers are deep learning models that use self-attention to understand
relationships in sequential data. They replaced RNNs and LSTMs in many tasks 
because they process all words at once instead of one-by-one.
"""

question = "Why are Transformers better than RNNs?"

result = question_answerer({
    "context": context,
    "question": question
})

print(result)
```
