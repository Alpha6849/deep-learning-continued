```python
# Create a sentiment analysis pipeline

sentiment_model = pipeline("sentiment-analysis")

# Test it with a sample sentence
result = sentiment_model("I really love how Transformers work!")

print(result)
```

```python
# Create a GPT-2 text generation pipeline
generator = pipeline("text-generation", model="gpt2")

result = generator(
    "Once upon a time in a small village,",
    max_length=50,      # how long the output should be
    num_return_sequences=1  # number of generated sentences
)

print(result[0]["generated_text"])
```

```python
# Create a question-answering pipeline
question_answerer = pipeline("question-answering")

# context to check (already built in model will cheeck on thsi though )
context = """
Transformers are deep learning models that use self-attention to understand
relationships in sequential data. They replaced RNNs and LSTMs in many tasks 
because they process all words at once instead of one-by-one.
"""

question = "Why are Transformers better than RNNs?"

result = question_answerer({
    "context": context,
    "question": question
})

print(result)
```

```python
# Create a Named Entity Recognition (NER) pipeline
ner_model = pipeline("ner", grouped_entities=True)

# Test it on a sample sentence
text = "Apple is releasing a new iPhone in California next month."

result = ner_model(text)

print(result)
```

```python
summarizer = pipeline("summarization", model="t5-small")

long_text = """
Transformers are neural network architectures that rely on self-attention 
mechanisms. They are faster to train and allow for parallel processing,
making them very effective for NLP tasks.
"""

summary = summarizer(long_text, max_length=40, min_length=10)

print(summary[0]['summary_text'])

```

```python
# Zero-shot classification pipeline
classifier = pipeline("zero-shot-classification", model="facebook/bart-large-mnli")

text = "I want to learn how to build transformers and deep learning models."

labels = ["education", "sports", "technology", "cooking"]

result = classifier(text, labels) # label classification without any training

print(result)
```
